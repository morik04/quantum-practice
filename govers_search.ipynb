Implement Quantum Methods

```python
oracle=QuantumCircuit(2, 0, name="oracle")
oracle.cz(0,1)
oracle.to_gate()
oracle.draw("mpl")
```

```python
grovers_circuit=QuantumCircuit(2, 2)
grovers_circuit.h([0, 1]) #applies to both qubits at once
grovers_circuit.append(oracle, [0, 1])
grovers_circuit.draw("mpl")
```

```python
reflection=QuantumCircuit(2, name="reflection")
reflection.h([0, 1])
reflection.z([0, 1])
reflection.cz(0, 1)
reflection.h([0, 1])
reflection.to_gate()
reflection.draw("mpl")
```

```python
grovers_circuit=QuantumCircuit(2, 2)
grovers_circuit.h([0, 1])
grovers_circuit.append(oracle, [0, 1])
grovers_circuit.append(reflection, [0, 1])
grovers_circuit.measure([0, 1], [0, 1])
grovers_circuit.draw("mpl")
```
