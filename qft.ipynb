Quantum Fourier Transform

```python
from qiskit import QuantumCircuit
from qiskit.circuit.library import UnitaryGate
import numpy as np
```

```python
angles=[]
for k in range(2, 11):
    phi=2/(2**k)
    angles.append(phi)
```

```python
#make function qft(n) that automatically makes a qft of n qubits

#add swaps
qc=QuantumCircuit(10, 0)
#first qubit
for k in range(0, 10):
    qc.h(k)
    
    for i in range(9-k):
        qc.crz(angles[i], k, k+i+1)
    qc.barrier()
    
qc.draw("mpl")
```
