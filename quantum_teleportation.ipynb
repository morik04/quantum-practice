Quantum Teleporatation

```python
from qiskit import QuantumCircuit as qc
import numpy as np
import matplotlib.pyplot as plt
```

```python
circuit1=qc(3, 3)
## Put q0 in custom state###
circuit1.y(0)


###
circuit1.h(1)
circuit1.cx(1, 2)
circuit1.cx(0, 1)
circuit1.h(0)
circuit1.measure(1, 0)
circuit1.cx(1, 2)
circuit1.measure(0, 1)
circuit1.cz(0, 2)
circuit1.measure(2, 2)
circuit1.draw('mpl')
```

```python
from qiskit_aer import AerSimulator
from qiskit.result import marginal_counts
```

```python
simulator = AerSimulator()
shots=10

job = simulator.run(circuit1, shots=shots)   #run on sim
full_result = job.result()
qubits_of_interest=[2]
marginal_counts_result = marginal_counts(full_result, indices=qubits_of_interest)
print(marginal_counts_result.get_counts())
```
