Quantum Fourier Transform

```python
from qiskit import QuantumCircuit
from qiskit.circuit.library import UnitaryGate
import numpy as np
```

```python
angles=[]
for k in range(2, 11):
    phi=2*np.pi/(2**k)
    angles.append(phi)
```

```python
qc=QuantumCircuit(10, 0)
for k in range(0, 10):
    qc.h(k)
    
    for i in range(9-k):
        qc.cp(angles[i], k, k+i+1)
    qc.barrier()  #makes it easier to read circuit
    
for i in range(5):
    qc.swap(i, 9-i)
    
# qc.draw("mpl")
```

```python
def QFT(n):
    qc=QuantumCircuit(n, 0)
    
    angles=[]
    for k in range(2, n+1):
        phi=2*np.pi/(2**k)
        angles.append(phi)
    for k in range(0, n):
        qc.h(k)
    
        for i in range(n-1-k):
            qc.cp(angles[i], k, k+i+1)
        qc.barrier()
        
    for i in range(n//2):
        qc.swap(i, n-i-1)
    return qc    
    
```

```python
qc4=QFT(5)
qc4.draw("mpl")
```
